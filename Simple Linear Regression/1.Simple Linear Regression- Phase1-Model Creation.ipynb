```python
""" Phase 1 : Step 1: 
Standard method to import Pandas library into the code to use 
for data analysis and manipulation 
Pandas library is loaded into current script's memory. 
Once imported we can gain access to all of its powerful data structures
pd is the alias name for pandas library. 
"""
import pandas as pd
```

```python
""" Step 2: 
read_csv is pandas built in function designed specifically to read CSV (Comma Seperated Values) files.
It takes teh file and automatically converts it into a Data Frame.
Salary_Data.csv is the name of the input file. Since no path is specified, 
Python expectst this to be located in the same folder as Python Script of Jupyter Notes 
Dataset= This is the assignment step where the entire contents of the fiel is stored in a 
variable called Dataset
"""
dataset=pd.read_csv("Salary_Data.csv")
```

```python
dataset
```

```python
""" Step 3:
To retrieve the input column into a variable we give 
input variable = dataframe[["input column name""]
"""
independent=dataset[["YearsExperience"]]
independent
```

```python
""" Step 4: 
To retrieve the output column into a variable we give 
output variable = dataframe[["output column name""]
"""
dependent=dataset[["Salary"]]
dependent
```

```python
#So with step 3 & step4 , we have split both input and output separately
#Now we have to separate training set and test set
```

```python
""" Step 5 : 
sklearn (scikit-learn) is the most popular machine learning library in Python
Model_Selection is s specific module inside which it has tools for evaluating and tuning machine learning  models
We fetch only train_test_split instead of importing entire Model_Selection
train_test_split shuffles and chops single dataset into two pieces : train set(70% - 80%) and test set(30% - 20%)

The output of train_test_split function returns 4 parameters - x_train, x_test, y_train and y_test
x_train and X_test - refers to input and y_train and y_test refers to output

"""
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(independent,dependent,test_size=0.30,random_state=0)
```

```python
""" Step 6 : 
The below two lines are procedures 
from sklearn.linear_model import LinearRegression
regressor=LinearRegression()
fit method (function) substitutes the values in the formula.
weight and bias would have been calculated once the fit function is executed.
Model is ready once the weight and bias are found out
"""
from sklearn.linear_model import LinearRegression
regressor=LinearRegression()
regressor.fit(x_train,y_train)
```

```python
""" Step 7 : 
Store the weight 
"""
weight=regressor.coef_
weight
```

```python
""" Step 8 : 
Store the bias  
"""
bias=regressor.intercept_
bias
```

```python
""" Step 9 : 
Evaluate the model  cross check the result 
"""
y_pred = regressor.predict(x_test)
```

```python
""" Step 10 :
r square - (Coefficient of Determination) is the standard metric used to evaluate how 
well a regression model performed
r_score nearly to one emans performing  very well
if it is near to zero, it is not performing well
"""
from sklearn.metrics import r2_score
r_score=r2_score(y_test,y_pred)
r_score
```

```python
""" Step 11: 
we have to save the model once it is confirmed that model is perfroming well

import pickle : 
Python's standard pickle module is loaded .
Pickling means serialization - 
converting a complex Python object (like a machine learning model, with all its internal 
weights and math formul) into a raw stream of bytes that can be saved to a disk file

filename="finalized_model_linear.sav"
.sav or .pkl are the standard file extensions used for saving machine learning models 
pickle.dump(regressor,open(filename,'wb'))

pickle.dump(regressor,open(filename,'wb'))
The above command actually writes the model to hard drive. It has two steps : 

a. open(filename,'wb') 
This opens the file for writing in binary mode - wb .Since machine language model is not 
regular text, it must be written as binary code.

b. pickle.dump(regressor,....)
This takes teh trained regressor model object,  converts into bytes, and dumps into 
that open binary file

we can find a new file with .sav extesnion in the project folder.
This is used for deployment. It also allows you to make instant predictions
on new data  without needing the original dataset or spending time retraining the model
"""
import pickle 
filename="finalized_model_linear.sav"
pickle.dump(regressor,open(filename,'wb'))
```

```python
""" Step 12 : 
Load machine learning model in Python 
open("finalized_model_linear.sav",'rb')
This opens the file containing your model. 'rb' stands for read binary, which is 
required as the model was saved as binary
pickle.load() 
This reads the binary stream and converts it back into a fully functioning Pandas/scikit-learn object
Loaded_model= This stores the restored model into a new variable name.
Loaded_model now possess the exact same patterns, weights and formulas it learned during its original 
training session.
result=loaded_model.predict([[15]])
the above line is used to predict the salary for 15 years of experience
"""
loaded_model=pickle.load(open("finalized_model_linear.sav",'rb'))
result=loaded_model.predict([[15]])
```
